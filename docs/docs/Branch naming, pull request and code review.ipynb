
# PentAI Contribution Guidelines

## 1. Purpose

These guidelines define how team members contribute to the
Livestock Disease Detection project using GitHub.

## 2. Branch Naming

Use `type/short-description`.

- `feature/` - new functionality
- `fix/` - bug fixes
- `docs/` - documentation
- `test/` - tests
- `refactor/` - code improvements without intended behavior changes
- `hotfix/` - urgent fixes

Examples:
- `feature/cattle-image-preprocessing`
- `feature/goat-disease-classifier`
- `fix/missing-image-validation`
- `docs/update-dataset-inventory`
- `test/model-evaluation`
- `refactor/image-loading`

Use lowercase letters and hyphens. Create one branch per task.
Include the GitHub Issue number when useful, such as
`feature/42-cattle-image-preprocessing`.

## 3. Before Starting

1. Check the team's ClickUp task and related GitHub Issues.
2. Synchronize your local default branch.
3. Create a task-specific branch from the default branch.
4. Avoid duplicating another member's work.

## 4. Development and Data

- Keep changes focused on the assigned task.
- Test code and notebooks before requesting review.
- Do not commit credentials, access tokens, or passwords.
- Do not commit large raw datasets or model files unless
  the team has explicitly agreed to track them.
- Follow the repository's existing data storage and .gitignore rules.
- Document changes to preprocessing, model training, and evaluation.
- Do not present model predictions as confirmed veterinary diagnoses.

## 5. Pull Requests

All normal changes must be submitted through a pull request
targeting the repository's default branch.

Each PR must explain:
- What changed
- Why the change was needed
- How it was tested
- Any known limitations
- The related issue, if applicable

Use a closing reference such as `Closes #42` when completing
a tracked issue.

## 6. Code Review

- Require at least one approval from another team member.
- Authors must not approve their own PRs.
- Review code correctness, readability, testing, and unintended changes.
- Check data leakage risks and evaluation methodology where relevant.
- Address blocking review feedback before merging.
- Resolve review conversations before merging when required.
- Re-request review after substantial changes.

## 7. Merge Policy

- Use Squash and merge by default.
- Merge only after required approvals and checks pass.
- Delete the task branch after a successful merge.
- Keep the default branch stable and usable.

## 8. Team Responsibilities

Contributors develop, test, document, and explain their changes.

Reviewers provide constructive feedback and check the changes
against project requirements.

The project maintainer manages repository settings and helps
resolve workflow or merge issues.

## 9. Collaboration

Use ClickUp for task assignment and progress tracking.
Use GitHub Issues for trackable technical tasks and bugs.
Use pull requests for discussion, review, and merging.

Ask the project maintainer if you need help with permissions,
merge conflicts, or the contribution process.
